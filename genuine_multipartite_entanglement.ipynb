# Notas sobre emaranhamento multipartite

Tais notas são baseadas nas Refs. [1,2].

## Emaranhamento tripartite

### Estados puros

__Definição__: Seja $\mathcal{H} \equiv \mathcal{H}_{ABC} \simeq \mathcal{H}_A \otimes \mathcal{H}_B  \otimes \mathcal{H}_C$ um espaço de Hilbert composto de 3 partes. $|\Psi\rangle_{ABC} \in \mathcal{H}$ é dito _completamente separável_ se pode ser escrito como:
$$|\Psi\rangle_{ABC} = |\psi \rangle_A \otimes | \phi \rangle_B \otimes | \chi \rangle_C \equiv |\Psi \rangle_{A|B|C},$$
onde $|\psi \rangle_A \in \mathcal{H}_A$ e assim por diante.

__Definição__: $|\Psi\rangle_{ABC} \in \mathcal{H}$ é dito _bi-separável_ se pode ser escrito como:
$$|\Psi\rangle_{ABC} = |\psi \rangle_A \otimes | \phi \rangle_{BC}\equiv |\Psi \rangle_{A|BC} \equiv |\Psi \rangle_{\text{bi-sep}},$$
onde $|\psi \rangle_A \in \mathcal{H}_A$ e $|\phi \rangle_{BC} \in \mathcal{H}_{BC}$. Notemos que o estado $|\phi \rangle_{BC}$ pode corresponder a um estado emaranhado entre as partes $B$ e $C$.

__Definição__: Qualquer estado puro $|\Psi\rangle_{ABC} \in \mathcal{H}$ que não é bi-separável é dito _genuinamente emaranhado de 3 partes_.

__Exemplos__ de estados genuinamente emanhados: 
\begin{align}
& |GHZ \rangle = \frac{1}{\sqrt{2}}\left( |0,0,0\rangle + |1,1,1\rangle\right), \\
& |W\rangle = \frac{1}{\sqrt{3}}\left(|1,0,0\rangle + |0,1,0\rangle\ + |0,0,1 \rangle \right).
\end{align}

Notemos que, para o estado GHZ,
\begin{align}
 \rho_{AB} = \text{Tr}_C |GHZ\rangle \langle GHZ | = \frac{1}{2}(|0,0\rangle \langle 0,0| + |1,1\rangle \langle 1,1|),
 \ \ \ \rho_A = \text{Tr}_{BC} |GHZ\rangle \langle GHZ | = \frac{1}{2}(|0\rangle \langle 0| + |1\rangle \langle 1|).
\end{align}
Ou seja, o estado reduzido $\rho_{AB}$ é separável e misto e o estado reduzido $\rho_A$ é maximamente misto. Logo, o estado GHZ não é robusto frente a perda de uma partição, no sentido de perder o emaranhamento se qualquer parte é descartada.

 
Notemos que, para o estado W,
\begin{align}
 \varrho_{AB} = \text{Tr}_C |W\rangle \langle W | = \frac{1}{3}|0,0\rangle \langle 0,0| + \frac{2}{3}|\Psi_+ \rangle \langle \Psi_+|,
 \ \ \ \varrho_A = \text{Tr}_{BC} |W\rangle \langle W | = \frac{2}{3}|0\rangle \langle 0| + \frac{1}{3}|1\rangle \langle 1|,
\end{align}
onde $|\Psi_+ \rangle = \left(|0,1\rangle + |1,0\rangle\right)/\sqrt{2}$. Ou seja, o estado reduzido $\varrho_{AB}$ é misto mas não separável pelo critério PPT e o estado reduzido $\rho_A$ é misto. Logo, o estado W é robusto frente a perda de uma partição, no sentido de ainda conter emaranhamento se uma das partes é descartada.

Tal observação levou a um resultado importante denominado de _Three qubits can be entangled in two inequivalent ways_ segundo a Ref.[3]. 

Para isso:

__Definição__: Quaisquer dois estados $|\Psi\rangle_{ABC}$ e $\Phi\rangle$ em $\mathcal{H}$ são ditos equivalentes frente à SLOCC (stochastic local operation and classical comunication, que correspondem à operações unitárias locais) se existem operadores unitários locais $A \in B(\mathcal{H}_A)$, $B \in B(\mathcal{H}_B)$ e $C \in B(\mathcal{H}_C)$ tais que
$$|\Psi\rangle = A \otimes B \otimes C |\Phi \rangle.$$


__Teorema__: Existem duas classes de equivalência diferentes de estados genuinamente emaranhados de 3 partes, i.e., 
$$\nexists \ A, B, C \ \ \ \text{tal que} \ \ \ |GHZ\rangle = A \otimes B \otimes C |W \rangle.$$



 


### Estados mistos

__Definição:__ O estado misto $\rho_{ABC}$ é dito _completamente separável_ se pode ser escrito como uma combinação convexa de estados puros completamente separáveis, i.e.,
$$\rho_{ABC} = \sum_i p_i |\psi_i\rangle_A \langle \psi_i |\otimes |\phi_i\rangle_B \langle \phi_i |\otimes |\chi_i\rangle_C \langle \chi_i| \equiv \rho_{A|B|C}. $$

__Definição:__ O estado misto $\rho_{ABC}$ é dito _separável por partição_ (partition-separable) se pode ser escrito como uma combinação convexa de estados puros que são separáveis em relação a uma _partição fixa_ (A|BC, ou AB|C, ou AC|B), i.e.,
$$\rho_{ABC} = \sum_i p_i |\psi_i\rangle_A \langle \psi_i| \otimes |\phi_i\rangle_{BC} \langle \phi_i |\equiv \rho_{A|BC}.$$

__Definição:__ O estado misto $\rho_{ABC}$ é dito _biseparável_ se pode ser escrito como uma combinação convexa de estados puros biseparáveis, i.e.,
$$\rho_{ABC} = \sum_i p_i |\psi_i\rangle_{\text{bi-sep}} \langle \psi_i | \equiv \rho_{\text{bi-sep}}.$$
__Observações:__ (i) Notemos que, para o estado biseparável, a partição não é fixa; (ii) as definições de separável por partição e biseparável coincidem para estados puros. (iii) Para qualquer estado separável por partição, a separabilidade de qualquer um de seus subsistemas com respeito à mesma bipartição é garantida. Por exemplo, um estado na bipartição AB|C permanece separável em relação a A|C se o subsistema B for traçado. No entanto, o inverso não é verdadeiro, em geral, para estados biseparáveis. A separabilidade dos estados reduzidos em relação a, por exemplo, A|C e B|C não implica a separabilidade em relação a AB|C.

__Definição:__ Qualquer estado misto $\rho_{ABC}$ que não é _biseparável_ é dito _genuinamente emaranhado de 3 partes_.

$$A|B|C \subset A|BC \cap AB|C \cap AC|B \subset A|BC, AB|C, AC|B \subset \text{bi-sep}.$$


__Exemplo:__ Consideremos o seguinte estado
\begin{align}
\rho_{ABC} = \frac{1}{3}\left(|\Psi_+\rangle_{AB} \langle \Psi_+| \otimes |0\rangle_C \langle 0| + |\Psi_+\rangle_{AC} \langle \Psi_+| \otimes |0\rangle_B \langle 0| + |\Psi_+\rangle_{BC} \langle \Psi_+| \otimes |0\rangle_A \langle 0| \right).
\end{align}
Tal estado corresponde a um exemplo de estado biseparável e, portanto, não genuinamente emaranhado. Além disso, tal estado não é separável por partição dado qualquer partição fixa. De fato, se fixarmos uma partição, digamos $AB|C$, pelo critério PPT, tal estado é emaranhado. Por fim, se traçarmos qualquer uma das partições, digamos $\rho_{AB} = \text{Tr}_C \rho_{ABC} $, o estado reduzido bipartido é separável. Ou seja, neste caso vemos que, embora a partição $AB = A|B$ é separável, o estado global $ABC = AB|C$ é emaranhado dado uma partição fixa, mas ainda sim, não é genuinamente emaranhado de 3 partes.



## Emaranhamento multipartite

### Estados puros
__Definição__ Seja $\mathcal{H} \simeq \otimes_{i = 1}^N \mathcal{H}_i = \mathcal{H}_1 \otimes ... \otimes \mathcal{H}_N$ o espaço de Hilbert composto de $N$ partes. Um estado $|\Phi\rangle \in \mathcal{H}$ é dito $k$-separável, para $1 < k \le N$, com respeito a partição $\mathcal{A}_1 | \mathcal{A}_2| ... | \mathcal{A}_k$ das N partes em k subconjuntos, se pode ser escrito como
\begin{align}
|\Phi\rangle = \otimes_{i = 1}^k | \phi_i \rangle_{\mathcal{A}_i} = |\phi_1 \rangle_{\mathcal{A}_1} \otimes ... \otimes |\phi_k \rangle_{\mathcal{A}_k} \equiv |\Phi\rangle_{\text{k-sep}},
\end{align}
onde $ | \phi_i \rangle_{\mathcal{A}_i} \in \mathcal{H}_{\mathcal{A}_i} $ e $\mathcal{H} \simeq \otimes_{i = 1}^k \mathcal{H}_{\mathcal{A}_i}$. 

Notemos que $\mathcal{H}$ é separável com respeito a k-partição $\{\mathcal{A}_1, \mathcal{A}_2, ... , \mathcal{A}_k \} \equiv \mathcal{A}_1 | \mathcal{A}_2| ... | \mathcal{A}_k$, onde
$$\mathcal{A}_i \subset \{1,2,...,N\}, \ \ \ \cup_{i = 1}^k \mathcal{A}_i  = \{1,2,...,N\}, \ \ \ \ \mathcal{A}_i \cap \mathcal{A}_j =  \emptyset, \ \forall i \neq j.$$

__Observação:__ A definição acima contém a definição de completamente separável $(k = N)$ assim como a definição de biseparabilidade $(k=2)$.


__Definição__: Qualquer estado puro $|\Psi\rangle \in \mathcal{H}$ que não é $k$-separável para todo $k \ge 2$ é dito _genuinamente emaranhado de N partes_.


### Estados mistos



__Definição:__ O estado misto $\rho \equiv \rho_{1 \ ... \ N}$ é dito _separável por partição_ (partition-separable) se pode ser escrito como uma combinação convexa de estados puros que são separáveis em relação a uma _partição fixa_ das N partes. Por exemplo, podemos considerar $\mathcal{A}_1 = \{1,2\}$, $\mathcal{A}_2 = \{3\}$,..., $\mathcal{A}_k=\{N-1,N\}$ de modo que
$$\rho = \sum_i p_i |\psi_i\rangle_{\mathcal{A}_1} \langle \psi_i| \otimes |\phi_i\rangle_{\mathcal{A}_2}\langle \phi_i |\otimes...\otimes |\chi\rangle_{\mathcal{A}_k}\langle \chi| =  \sum_i p_i |\psi_i\rangle_{12} \langle \psi_i| \otimes |\phi_i\rangle_{3}\langle \phi_i |\otimes...\otimes |\chi\rangle_{N-1 \ N}\langle \chi| \equiv \rho_{\mathcal{A}_1|\mathcal{A}_2| ...|\mathcal{A}_k}.$$

__Definição:__ O estado misto $\rho$ é dito _$k$-separável_ (sem partição fixa) se pode ser escrito como uma combinação convexa de estados puros $k$-separáveis, i.e.,
$$\rho = \sum_i p_i |\psi_i\rangle_{\text{k-sep}} \langle \psi_i | \equiv \rho_{\text{k-sep}}.$$

__Observação:__ A definição acima contém a definição de completamente separável $(k = N)$ assim como a definição de biseparabilidade $(k=2)$. Além disso, Embora a k-separabilidade implique que existe uma decomposição na qual cada termo é um produto tensorial de k subsistemas, isso não significa que cada termo precise ser um produto tensorial de exatamente k subsistemas, desde que seja um produto tensorial de pelo menos k subsistemas. Por exemplo, Notemos que um estado misto bisseparável (2-sep), em um espaço de Hilbert de $N = 4$ partes, pode ser escrito como uma mistura convexa de termos que são separáveis em relação às bipartições $A∣BCD$,..., mas também relação às bipartições $AB∣CD$,... Além disso, também podem estar presentes termos separáveis em relação às tripartições $AB∣C∣D$,..., ou até mesmo termos totalmente separáveis $A∣B∣C∣D$. Por fim, um estado k-separável ainda pode conter emaranhamento (genuinamente multipartido) entre até $N−k+1$ partes. 

__Definição:__ Qualquer estado misto $\rho$ que não é _$k$-separável_ para todo $k \ge 2$ é dito _genuinamente emaranhado de N partes_. Ou seja, $|\Psi\rangle \in \mathcal{H}$ é genuinamente emaranhado para $k = 1$, i.e., para todas as partições possíveis do sistema há emaranhamento.

De fato, basta não ser $k = 2$ separável, pois
\begin{align}
N\text{-sep} \ \subset \ (N-1)\text{-sep} \ \subset ... \subset \ 3\text{-sep} \subset 2\text{-sep}.
\end{align}

__Observação__ Notemos também que quando comparamos uma combinação convexa de estados $\rho_{AB|CD}$ and $\rho_{AC|BD}$ com a combinação convexa de estados $\rho_{A|BCD}$ e $\rho_{ABC|D}$. Ambos são estados biseparáveis, no entanto o primeiro pode apresentar emaranhamento entre duas partes enquanto o último pode ainda apresentar emaranhamento genuino de 3 partes.

Assim, como notado na Ref.~[4], no limite termodinâmico $N >> 1$, a noção de $k$-separabilidade para o sistema total torna-se problemática. Isso se deve principalmente a dois motivos. Primeiro, essa noção não leva em consideração quantas partículas estão emaranhadas. Um estado biseparável pode ser um estado produto para duas partições de mesmo tamanho, ou apenas um único qubit pode estar separado de um grande estado genuinamente emaranhado multipartite; esses casos não são diferenciados. Segundo, afirmações sobre $k$-separabilidade exigem o conhecimento exato de $N$, uma condição prévia que geralmente não é satisfeita em situações realistas. Em contraste marcante, a noção de $k$-produtibilidade foi concebida para ser sensível à questão: “Quantos qubits estão emaranhados?”

Tal pergunta levou a seguinte definição de $k$-produtibilidade pelos autores da Ref.~[3].

__Definição__ Um estado puro $|\psi\rangle \in \mathcal{H}$ é dito _produzível por emaranhamento de $k$-partes_ (ou só _$k$-produzível_) se pode ser escrito na forma 
$$ |\psi \rangle = |\phi_1 \rangle_{\mathcal{A}_1} \otimes |\phi_2 \rangle_{\mathcal{A}_2} \otimes ... \otimes | \phi_m \rangle_{\mathcal{A}_m} \equiv |\psi \rangle_{\text{k-prod}}, $$
onde $|\phi_i \rangle_{\mathcal{A}_i}$ são estados emaranhados de no máximo k-partes e com $m \ge N/k$.

Assim, se o estado $|\psi\rangle \in \mathcal{H}$ é $k$-produzível então tal estado não corresponde a um estado genuinamente emaranhado de $k + 1$ partes.

__Definição:__ O estado misto $\rho$ é dito _$k$-produzível_ se pode ser escrito como uma combinação convexa de estados puros $k$-produtíveis, i.e., uma soma convexa de estados produto na qual cada fator tensorial é um estado de no máximo $k$ partes.
$$\rho = \sum_i p_i |\psi_i\rangle_{\text{k-prod}} \langle \psi_i | \equiv \rho_{\text{k-prod}}.$$

Um estado $N$-separável (completamente separável) é $1$-produzível, enquanto um estado genuinamente emaranhado de $N$ partes é $N$-produzível, mas não é $(N−1)$-produzível e, portanto, possui uma profundidade de emaranhamento igual a $N$. De maneira geral, um estado que não é $k$-produzível corresponder a um estado genuinamente emaranhado de pelo menos $k+1$-partes.


#### Representação operacional de um estado biseparável e sua purificação

##### Maneira ingênua
Notemos que um estado biseparável pode ser escrito como [5]:
$$
\rho_{\text{bi-sep}} = \sum_i p_i  \otimes_{j = 1}^2 \left(| \psi_{i(j)} \rangle_{\mathcal{A}_{i(j)}} \langle \psi_{i(j)}|\right) = \sum_i p_i  | \psi_{i(1)} \rangle_{\mathcal{A}_{i(1)}} \langle \psi_{i(1)}| \otimes | \psi_{i(2)} \rangle_{\mathcal{A}_{i(2)}} \langle \psi_{i(2)}|.
$$
Uma purificação possível é
$$
|\Psi(\rho_{\text{bi-sep}})\rangle  = \sum_{i} \sqrt{p_i}  \otimes_{j = 1}^2 \left(| \psi_{i(j)} \rangle_{\mathcal{A}_{i(j)}} \otimes |b_{ij}\rangle_B\right) = \sum_{i} \sqrt{p_i} \left(| \psi_{i(1)} \rangle_{\mathcal{A}_{i(1)}} \otimes |b_{i1}\rangle_{B_1}\right) \otimes \left(| \psi_{i(2)} \rangle_{\mathcal{A}_{i(2)}} \otimes |b_{i2}\rangle_{B_2}\right).
$$

__Observação__:No entanto, pela definição da Ref.[1], cada termo de um estado biseparável é o produto tensorial de pelo menos 2 subsistemas. Ou seja, em cada termo pode conter tripartições (k = 3), quadripartições (k = 4) e assim sucessivamente, escondido dentro dentro da bipartição. Ou seja,
$$
\rho_{\text{bi-sep}} = \sum_i p_i  \otimes_{j = 1}^2 \left(| \psi_{i(j)} \rangle_{\mathcal{A}_{i(j)}} \langle \psi_{i(j)}|\right) + \sum_i q_i  \otimes_{j = 1}^3 \left(| \phi_{i(j)} \rangle_{\mathcal{A}_{i(j)}} \langle \phi_{i(j)}|\right) + ... + \sum_i r_i  \otimes_{j = 1}^k \left(| \chi_{i(j)} \rangle_{\mathcal{A}_{i(j)}} \langle \chi_{i(j)}|\right) + ... 
$$


__Observação__: Notemos que, dado uma bipartição $\mathcal{A} = \{\mathcal{A}_1,\mathcal{A}_2\}$ de um sistema de N partes, podemos montar $|\mathcal{A}| = 2^{N-1} - 1$ combinações possíveis das diferentes partes do sistema. Por exemplo, se considerarmos um estado de 3 partes, então $\mathcal{A}_i \subset \{1,2,3\}$, de modo que há 3 bipartições possíveis: $1|23,12|3,13|2$. Já, para um estado de 4 partes, há 7 partições possíveis. E assim por diante. No entanto, para um estado ser biseparável, não é necessário que todas as combinações apareçam. Além disso, como mencionado no item anterior, pode haver termos com $k>2$-partições.


##### Maneira interessante  
Uma outra maneira de representar um estado separável é seguindo as Refs.[6, 7]. Dado uma bipartição fixa $\mathcal{A} = \{\mathcal{B}, \bar{\mathcal{B}}\}$, onde $\bar{\mathcal{B}}$ é o complemento de $\mathcal{B}$. Consideremos os estados separáveis por bipartição fixa representados por 
$$\rho_{\mathcal{B}|\bar{\mathcal{B}}} = \sum_j p_j |\phi_j\rangle_{\mathcal{B}}\langle \phi_j| \otimes |\chi_j\rangle_{\bar{\mathcal{B}}}\langle \chi_j|.$$
Assim, os estados biseparáveis são obtidos considerando combinações convexas das diferentes possiveis bipartições, i.e.,
$$
\rho_{\text{bi-sep}} = \sum_{\mathcal{B}} q_{\mathcal{B}} \rho_{\mathcal{B}|\bar{\mathcal{B}}} = \sum_{\mathcal{B}} q_{\mathcal{B}} \left(\sum_j p^{\mathcal{B}}_j |\phi_j\rangle_{\mathcal{B}}\langle \phi_j| \otimes |\chi_j\rangle_{\bar{\mathcal{B}}}\langle \chi_j| \right),
$$
de modo que a soma $\sum_{\mathcal{B}}$ irá conter no máximo $2^{N-1} - 1$. Note também que os termos com $k>2$-partições estão contidos dentro de cada $\rho_{\mathcal{B}|\bar{\mathcal{B}}}$.

Com isso, uma purificação possível para o estado acima é dada por
$$
|\Psi(\rho_{\text{bi-sep}})\rangle = \sum_{\mathcal{B}}\sum_j \sqrt{q_{\mathcal{B}} p^{\mathcal{B}}_j} |\phi_j\rangle_{\mathcal{B}} \otimes |\chi_j\rangle_{\bar{\mathcal{B}}} \otimes |c_{j\mathcal{B}}\rangle_{\mathcal{C}}.
$$


##### Purificação em termos de unitárias

\begin{align}
|\Psi(\rho_{\text{bi-sep}})\rangle & = \sum_{\mathcal{B}}\sum_j \sqrt{q_{\mathcal{B}} p^{\mathcal{B}}_j} |\phi_j\rangle_{\mathcal{B}} \otimes |\chi_j\rangle_{\bar{\mathcal{B}}} \otimes |c_{j\mathcal{B}}\rangle_{\mathcal{C}} \\
& = U_{\mathcal{C}} \sum_{\mathcal{B}}\sum_j \sqrt{q_{\mathcal{B}} p^{\mathcal{B}}_j} |\phi_j\rangle_{\mathcal{B}} \otimes |\chi_j\rangle_{\bar{\mathcal{B}}} \otimes |j\rangle_{\mathcal{C}_1} \otimes |\mathcal{B}\rangle_{\mathcal{C}_2}\\
& =  U_{\mathcal{C}} \sum_{\mathcal{B}}\sum_j \sqrt{q_{\mathcal{B}} p^{\mathcal{B}}_j} U_j^{\mathcal{B}} |j\rangle_{\mathcal{B}} \otimes \bar{U}_j^{\bar{\mathcal{B}}} |j\rangle_{\bar{\mathcal{B}}} \otimes |j\rangle_{\mathcal{C}_1} \otimes |\mathcal{B}\rangle_{\mathcal{C}_2}\\
& = U_{\mathcal{C}}  C_{U_j^{\mathcal{B}} \ \bar{U}_j^{\bar{\mathcal{B}}}}^{\mathcal{C} \to \mathcal{B},\bar{\mathcal{B}}}  \sum_{\mathcal{B}}\sum_j \sqrt{q_{\mathcal{B}} p^{\mathcal{B}}_j}  |0\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}} \otimes |j\rangle_{\mathcal{C}_1} \otimes |\mathcal{B}\rangle_{\mathcal{C}_2} \\
& = U_{\mathcal{C}} C_{U_j^{\mathcal{B}} \ \bar{U}_j^{\bar{\mathcal{B}}}}^{\mathcal{C} \to \mathcal{B},\bar{\mathcal{B}}} |0\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}} \otimes \sum_{\mathcal{B}}\sum_j \sqrt{q_{\mathcal{B}} p^{\mathcal{B}}_j}  |j\rangle_{\mathcal{C}_1} \otimes |\mathcal{B}\rangle_{\mathcal{C}_2} \\
& = U_{\mathcal{C}} C_{U_j^{\mathcal{B}} \ \bar{U}_j^{\bar{\mathcal{B}}}}^{\mathcal{C} \to \mathcal{B},\bar{\mathcal{B}}}  C_{V^{\mathcal{C_1}}(\theta_{\mathcal{B}})}^{\mathcal{C}_2 \to \mathcal{C}_1} V^{\mathcal{C_2}}(\phi) |0\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}} \otimes |0\rangle_{\mathcal{C}_1} \otimes |0\rangle_{\mathcal{C}_2},\\
& =  U_{\mathcal{C}} C_{U_j^{\mathcal{B}} \ \bar{U}_j^{\bar{\mathcal{B}}}}^{\mathcal{C} \to \mathcal{B},\bar{\mathcal{B}}} V_{\mathcal{C}} |0\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}} \otimes |0\rangle_{\mathcal{C}_1} \otimes |0\rangle_{\mathcal{C}_2},
\end{align} 
onde
$$
C_{U_j^{\mathcal{B}} \ \bar{U}_j^{\bar{\mathcal{B}}}}^{\mathcal{C} \to \mathcal{B},\bar{\mathcal{B}}} = \sum_{j, \mathcal{B}}U_j^{\mathcal{B}} \otimes \bar{U}_j^{\bar{\mathcal{B}}} \otimes |j\rangle_{\mathcal{C}_1} \langle j | \otimes |\mathcal{B}\rangle_{\mathcal{C}_2} \langle \mathcal{B}|.
$$


##### Exemplo

Consideremos o seguinte estado biseparável de tripartite
\begin{align}
\rho_{ABC} = \frac{1}{3}\left(|\Phi_+\rangle_{AB} \langle \Phi_+| \otimes |0\rangle_C \langle 0| + |\Phi_+\rangle_{AC} \langle \Phi_+| \otimes |0\rangle_B \langle 0| + |\Phi_+\rangle_{BC} \langle \Phi_+| \otimes |0\rangle_A \langle 0| \right),
\end{align}
onde 
$|\Phi_+ \rangle = \left(|0,0\rangle + |1,1\rangle\right)/\sqrt{2}$.


Como é sistema tripartite, dado a bipartição $\mathcal{A} = \{\mathcal{B}, \bar{\mathcal{B}}\}$, temos as seguintes 3 possibilidade
\begin{align}
& \mathcal{B} = AB, AC, BC \to 0,1,2 \\
& \bar{\mathcal{B}} = C, B, A.
\end{align}
ou seja, fazemos a codificação $AB \to 0, AC \to 1, BC \to 2$.

Além disso, dado uma bipartição fixa $\{\mathcal{B}, \bar{\mathcal{B}}\}$, temos que o estado para uma bipartição fixa é
$$
\rho_{\mathcal{B}|\bar{\mathcal{B}}} = |\Phi_+\rangle_{\mathcal{B}} \langle \Phi_+| \otimes |0\rangle_{\bar{\mathcal{B}}} \langle 0| = \sum_j \delta_{j0} |\Phi_+\rangle_{\mathcal{B}} \langle \Phi_+| \otimes |j\rangle_{\bar{\mathcal{B}}} \langle j|,
$$
ou seja, 
$$p_{j\mathcal{B}} = \delta_{j0}, \ \ q_{\mathcal{B}} = 1/3.$$

Com isso, temos a seguinte possível purificação
$$
|\Psi(\rho_{\text{bi-sep}})\rangle = \sum_{\mathcal{B} = 0}^2\sum_{j = 0}^1 \sqrt{\frac{1}{3}} \delta_{j0}|\Phi_+\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}} \otimes |j\rangle_{\mathcal{C}_1} \otimes |\mathcal{B} \rangle_{\mathcal{C}_2},
$$
com o sistema auxiliar $\mathcal{C}_1$ sendo um qubit e o sistema auxiliar $\mathcal{C}_2$ sendo um qutrit.

Em termos de unitárias, primeiro percebemos que
\begin{align}
 \sum_{\mathcal{B} = 0}^2 \sqrt{\frac{1}{3}} |\mathcal{B} \rangle_{\mathcal{C}_2} & \equiv  \sum_{j,k=0}^1 c_{jk}|jk\rangle_{DE} \\
& = C^{D_0\rightarrow E}_{R_y(\theta_0)}C^{D_1\rightarrow E}_{R_y(\theta_1)}(R^D_y(\xi)\otimes I)|00\rangle \\ 
& = \cos(\xi/2)\cos(\theta_0/2)|00\rangle + \cos(\xi/2)\sin(\theta_0/2)|01\rangle + \sin(\xi/2)\cos(\theta_1/2)|10\rangle + \sin(\xi/2)\sin(\theta_1/2)|11\rangle,
\end{align}
com
$$ \theta_0 = \frac{\pi}{2}, \ \ \theta_1 = 0, \ \ \xi = 2 \arccos(\sqrt{\frac{2}{3}}). $$
Logo,
$$ 
V^{\mathcal{C_2}}(\phi) = C^{D_0\rightarrow E}_{R_y(\theta_0)}C^{D_1\rightarrow E}_{R_y(\theta_1)}(R^D_y(\xi)\otimes I).
$$
com sistema $\mathcal{C_2}$ sendo codificado em dois qubits.

Em seguida, como
$$
\rho_{\mathcal{B}|\bar{\mathcal{B}}} = |\Phi_+\rangle_{\mathcal{B}} \langle \Phi_+| \otimes |0\rangle_{\bar{\mathcal{B}}} \langle 0| = \sum_j \delta_{j0} |\Phi_+\rangle_{\mathcal{B}} \langle \Phi_+| \otimes |j\rangle_{\bar{\mathcal{B}}} \langle j|,
$$
temos que
$$
C_{V^{\mathcal{C_1}}(\theta_{\mathcal{B}})}^{\mathcal{C}_2 \to \mathcal{C}_1}  = I_{\mathcal{C}_1} \otimes |0\rangle_{\mathcal{C}_2 } \langle 0| + I_{\mathcal{C}_1} \otimes |1\rangle_{\mathcal{C}_2 } \langle 1|  + I_{\mathcal{C}_1} \otimes |2\rangle_{\mathcal{C}_2 } \langle 2|   =  I_{\mathcal{C}_1}  \otimes  I_{\mathcal{C}_2} 
$$

Como comentado anterior, o sistema auxiliar $\mathcal{C}_1$ é um qubit. Como temos apenas um termo não nulo no somatório em $j$, sendo justamente o $j = 0$, as portas $C_{X}^{\mathcal{C}_1 \to \mathcal{B}}, C_{X}^{\mathcal{C}_1 \to \bar{\mathcal{B}}}$ serão equivalentes a identidades. 

Por fim, dado que
$$
|\Phi_+\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}} = \frac{1}{\sqrt{2}} \left(|0,0\rangle_{\mathcal{B}_1 \mathcal{B}_2} + |1,1\rangle_{\mathcal{B}_1 \mathcal{B}_2}\right)\otimes |0\rangle_{\bar{\mathcal{B}}} = C_{X}^{\mathcal{B}_1 \to \mathcal{B}_2} H_{\mathcal{B}_1} |0,0\rangle_{\mathcal{B}} \otimes |0\rangle_{\bar{\mathcal{B}}},
$$
então
$$
U_j^{\mathcal{B}} = \delta_{j0}  C_{X}^{\mathcal{B}_1 \to \mathcal{B}_2} H_{\mathcal{B}_1} = U_0^{\mathcal{B}}, \ \ \bar{U}_j^{\bar{\mathcal{B}}}  = \delta_{j0}  I_{\bar{\mathcal{B}}}.
$$
Logo,
$$
C_{U_j^{\mathcal{B}} \ \bar{U}_j^{\bar{\mathcal{B}}}}^{\mathcal{C} \to \mathcal{B},\bar{\mathcal{B}}} =  \sum_{\mathcal{B} = 0}^2 U_0^{\mathcal{B}} \otimes  I_{\bar{\mathcal{B}}}  \otimes |0\rangle_{\mathcal{C}_1} \langle 0 | \otimes |\mathcal{B}\rangle_{\mathcal{C}_2} \langle \mathcal{B}|.
$$
