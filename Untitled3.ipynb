```python
from functools import partial

import numpy as np
import scipy.optimize
import matplotlib.pyplot as plt


def z(x, y):
    
    gammaw = 1/np.sqrt(1 - x**2)
    gammaa = 1/np.sqrt(1 - y**2)

    D = np.sqrt(((gammaw + 1)/(gammaw - 1))*((gammaa + 1)/(gammaa - 1)))
    
    return D**2 - 2

x_window = 0.95, 0.9999
y_window = 0.95, 0.9999

xs = []
ys = []
for x in numpy.linspace(*x_window, num=200):
    try:
        # A more efficient technique would use the last-found-y-value as a 
        # starting point
        y = scipy.optimize.brentq(partial(z, x), *y_window)
    except ValueError:
        # Should we not be able to find a solution in this window.
        pass
    else:
        xs.append(x)
        ys.append(y)

plt.plot(xs, ys, color="red",  linewidth=1.5, linestyle="-.", label=r'$\phi = \frac{\pi}{2}$')
plt.xlim(*x_window)
plt.ylim(*y_window)
plt.xlabel(r'$v_{\alpha}$',fontsize = 16); plt.ylabel(r'$v_{\omega}$', fontsize=16)
plt.legend(fontsize=14)

plt.savefig('phi.pdf', format='pdf',dpi = 1200, bbox_inches='tight')

plt.show()
```
