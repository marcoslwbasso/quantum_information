# Giorgi State

Let $ |\Psi \rangle = \sqrt{p \epsilon} |0,0,0 \rangle + \sqrt{p(1- \epsilon)} |1,1,1 \rangle + \sqrt{(1 - p)/2}(|1,0,1 \rangle + |1,1,0 \rangle) $ such that $p,\epsilon \in [0,1]$. The reduced states $\rho_A$ incoherent and 

\begin{equation}
\rho_B = \rho_C = (p \epsilon + (1 - p)/2) |0 \rangle \langle 0 | + (p(1-\epsilon) + (1 - p)/2) |1 \rangle \langle 1 | + \sqrt(p(1-\epsilon)(1-p)/2)|0 \rangle \langle 1 | + h.c.
\end{equation}

#### Density matter for a Tolman-like distribution

Consideremos a seguinte distribuição de matéria: 
\begin{align}
& 8 \pi \rho = \frac{15 m}{r_0^3} \frac{r^4}{\Sigma^2}(1 - \frac{r^2}{r_0^2}) \Theta(r_0 -r), \\
\end{align}
onde $\Sigma = r^2 + a^2 \cos^2 \theta$. Vamos fazer $m = r_0 = 1$:

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
r = np.arange(0, 1, 0.01)
x = np.arange(0, np.pi, 0.01)
plt.style.use(['science', 'notebook', 'grid'])


r, x = np.meshgrid(r, x) # a = T_1, b = T_2



V = 15*r**4*(1 - r**2)/(r**2 + 0.7*(np.cos(x))**2)**2

plt.contourf(r,x,V, levels=100)
plt.colorbar(label='$density$')
plt.xlabel('$r$')
plt.ylabel('$theta$')
```
