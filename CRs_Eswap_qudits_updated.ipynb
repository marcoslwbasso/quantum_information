# Swap de emaranhamento para qudits

## The qudit algebra

Fourier gate
$$F = \frac{1}{\sqrt{d}}\sum_{j,k=0}^{d-1}e^{2\pi ijk/d}|j\rangle\langle k|;\ F|k\rangle = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{2\pi ijk/d}|j\rangle$$
State shift operator
$$X(j) = \sum_{k=0}^{d-1}|j\oplus k\rangle\langle k|;\ X(j)|k\rangle = |j\oplus k\rangle$$
Notação
$$j\oplus k = (j+k)\bmod d$$
Phase shift operator
$$Z(j) = \sum_{k=0}^{d-1}e^{2\pi ijk/d}|k\rangle\langle k|;\ Z(j)|k\rangle = e^{2\pi ijk/d}|k\rangle$$
Controll not
$$CNOT^{A\rightarrow B} = \sum_{j=0}^{d-1}|j\rangle\langle j|\otimes X(j);\ CNOT^{A\rightarrow B}|jk\rangle = |j\rangle|j\oplus k\rangle$$
Maximally entangle state
$$|\Phi\rangle = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}|jj\rangle$$
Heisenberg-Weyl operators
$$W_{jk} = X(j)Z(k);\ W_{jk}|l\rangle = e^{2\pi ikl/d}|j\oplus l\rangle$$
Bell basis (which is orthonormal and complete)
$$|\Phi_{jk}\rangle = (W_{jk}\otimes I)|\Phi\rangle = CNOT^{B\rightarrow A}(I\otimes F)|jk\rangle$$

## Qubits

In this case 
$$|\Phi\rangle = \sum_{j=0}^1|jj\rangle/\sqrt{2} = (|00\rangle+|11\rangle)/\sqrt{2}.$$
The Bell states are obtained as follows
\begin{align}
|\Phi_{jk}\rangle & = (W_{jk}\otimes I)|\Phi\rangle \\
& = (X(j)Z(k)\otimes I)\sum_{l=0}^1|ll\rangle/\sqrt{2} \\
& = \sum_{l=0}^1 X(j)Z(k)|l\rangle\otimes|l\rangle/\sqrt{2} \\
& = \sum_{l=0}^1 e^{2\pi ikl/2}X(j)|l\rangle\otimes|l\rangle/\sqrt{2} \\
& = \sum_{l=0}^1 e^{2\pi ikl/2}|(j+l)\bmod 2\rangle\otimes|l\rangle/\sqrt{2} \\
& = \big(e^{2\pi ik0/2}|(j+0)\bmod 2\rangle\otimes|0\rangle + e^{2\pi ik/2}|(j+1)\bmod 2\rangle\otimes|1\rangle\big) /\sqrt{2} \\
& = \big(|j\rangle\otimes|0\rangle + e^{\pi ik}|\overline{j}\rangle\otimes|1\rangle\big) /\sqrt{2} \\
& = \big(|j\rangle\otimes|0\rangle + e^{\pi ik}|j\oplus 1\rangle\otimes|1\rangle\big) /\sqrt{2}
\end{align}
So
\begin{align}
|\Phi_{00}\rangle & = \big(|0\rangle\otimes|0\rangle + e^{\pi i0}|1\rangle\otimes|1\rangle\big) /\sqrt{2} = |\Phi_+\rangle, \\
|\Phi_{01}\rangle & = \big(|0\rangle\otimes|0\rangle + e^{\pi i}|1\rangle\otimes|1\rangle\big) /\sqrt{2} = |\Phi_-\rangle, \\
|\Phi_{10}\rangle & = \big(|1\rangle\otimes|0\rangle + e^{\pi i0}|0\rangle\otimes|1\rangle\big) /\sqrt{2} = |\Psi_+\rangle, \\
|\Phi_{11}\rangle & = \big(|1\rangle\otimes|0\rangle + e^{\pi i}|0\rangle\otimes|1\rangle\big) /\sqrt{2} = -|\Psi_-\rangle.
\end{align}

In the context of the entanglement swapping experiment, we start with
\begin{align}
|\Psi\rangle & = |\xi\rangle_{AC}|\eta\rangle_{C'B} = \sum_{j,k=0}^1 c_{jk}|jk\rangle_{AC}\sum_{l,m=0}^1 d_{lm}|lm\rangle_{C'B} \\
& = \sum_{j,k,l,m=0}^1 c_{jk}d_{lm}|jm\rangle_{AB} |kl\rangle_{CC'}.
\end{align}

The post-BBM state is proportional to
\begin{align}
|\phi_{pq}\rangle_{ABCC'} & = (I_{AB}\otimes|\Phi_{pq}\rangle_{CC'}\langle\Phi_{pq}|)|\Psi\rangle \\
& = \sum_{j,k,l,m=0}^1 c_{jk}d_{lm}|jm\rangle_{AB} |\Phi_{pq}\rangle_{CC'}\langle\Phi_{pq}|kl\rangle_{CC'} \\
& = \sum_{j,k,l,m=0}^1 c_{jk}d_{lm}|jm\rangle_{AB} |\Phi_{pq}\rangle_{CC'}2^{-1/2}(\delta_{pk}\delta_{0l} + e^{-i\pi q}\delta_{\overline{p}k}\delta_{1l}) \\
& = \frac{1}{\sqrt{2}}\sum_{j,m=0}^1 \big(c_{jp}d_{0m}+c_{j\overline{p}}d_{1m}e^{-i\pi q}\big)|jm\rangle_{AB} |\Phi_{pq}\rangle_{CC'} \\
\end{align}
em que usamos
\begin{align}
\langle\Phi_{pq}|kl\rangle_{CC'} & = 2^{-1/2}(\langle p0|+e^{-i\pi q}\langle\overline{p}1|)|kl\rangle = 2^{-1/2}(\langle p0|kl\rangle+e^{-i\pi q}\langle\overline{p}1|kl\rangle) \\
& = 2^{-1/2}(\delta_{pk}\delta_{0l} + e^{-i\pi q}\delta_{\overline{p}k}\delta_{1l}).
\end{align}

Normalizing the vector we get
\begin{align}
|\hat{\phi}_{pq}\rangle_{AB} & = \frac{|\phi_{pq}\rangle_{AB}}{|||\phi_{pq}\rangle_{AB}||}.
\end{align}
We would like to look at the entanglement of this state. But it is too combersome. For the qudit case below, it is worse.

## Qudits

Consider two pairs of qudits, with dimension $d$, in arbitrary states
$$|\Psi\rangle = |\xi\rangle_{AC}|\eta\rangle_{C'B} = \sum_{j,k=0}^{d-1}c_{jk}|jk\rangle\sum_{l,m=0}^{d-1}d_{lm}|lm\rangle = \sum_{j,k,l,m=0}^{d-1}c_{jk}d_{lm}|jm\rangle_{AB}\otimes|kl\rangle_{CC'}.$$

Measure $CC'$ on the Bell basis. If the result $|\Phi_{jk}\rangle$ is obtained, the post-measurement state is
$$|\hat{\phi}_{pq}\rangle = \frac{|\phi_{pq}\rangle}{|||\phi_{pq}\rangle||} = \frac{(I_{AB}\otimes\Pi_{\Phi_{pq}})|\Psi\rangle}{||(I_{AB}\otimes\Pi_{\Phi_{pq}})|\Psi\rangle||}.$$
Besides
\begin{align}
|\phi_{pq}\rangle & = (I_{AB}\otimes\Pi_{\Phi_{pq}})|\Psi\rangle \\ 
& = \sum_{j,k,l,m=0}^{d-1}c_{jk}d_{lm}|jm\rangle_{AB}\otimes\langle \Phi_{pq}|kl\rangle |\Phi_{pq}\rangle_{CC'}.
\end{align}

It follows that
\begin{align}
|\Phi\rangle_{pq} & = (W_{pq}\otimes I)|\Phi\rangle = (X(p)Z(q)\otimes I)\frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}|jj\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}X(p)Z(q)|j\rangle\otimes|j\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{2\pi iqj/d}X(p)|j\rangle\otimes|j\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{2\pi iqj/d}|(p+j)\bmod d\rangle\otimes|j\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{2\pi iqj/d}|p\oplus j\rangle\otimes|j\rangle.
\end{align}
So
\begin{align}
\langle\Phi_{pq}|kl\rangle & = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{-2\pi iqj/d}(\langle(p+j)\bmod d|\otimes\langle j|)(|k\rangle\otimes|l\rangle) \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{-2\pi iqj/d}\langle(p+j)\bmod d|k\rangle\otimes\langle j|l\rangle = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{-2\pi iqj/d}\langle(p+j)\bmod d|k\rangle\delta_{jl} \\
& = \frac{1}{\sqrt{d}}e^{-2\pi iql/d}\langle(p+l)\bmod d|k\rangle \\
& = \frac{1}{\sqrt{d}}e^{-2\pi iql/d}\delta_{k,p\oplus l}
\end{align}

Thus
\begin{align}
|\phi_{pq}\rangle & = \sum_{j,k,l,m=0}^{d-1}c_{jk}d_{lm}|jm\rangle_{AB}\otimes\frac{1}{\sqrt{d}}e^{-2\pi iql/d}\delta_{k,p\oplus l} |\Phi_{pq}\rangle_{CC'} \\
& = \Big(\frac{1}{\sqrt{d}}\sum_{j,k,l,m=0}^{d-1}c_{jk}d_{lm}e^{-2\pi iql/d}\delta_{k,p\oplus l}|jm\rangle_{AB}\Big)\otimes |\Phi_{pq}\rangle_{CC'} \\
& = \Big(\sum_{j,m=0}^{d-1}\Big(\frac{1}{\sqrt{d}}\sum_{l=0}^{d-1}c_{j,p\oplus l}d_{lm}e^{-2\pi iql/d}|jm\rangle_{AB}\Big)\Big)\otimes |\Phi_{pq}\rangle_{CC'} \\
\end{align}
This state is also too cumbersome to analyze.

## Zero local coherence states are enough!

Let us argue that in the context of the entanglement swap protocol, as local unitary transformations do not change entanglement, we can use initial partially entangled states with null local coherence.

For general pure initial states
$$|\xi\rangle_{AC} = \sum_{j,k=0}^{d-1}c_{jk}|jk\rangle,\ |\eta\rangle_{C'B} = \sum_{j,k=0}^{d-1}d_{jk}|jk\rangle$$
we have
\begin{align}
& P(\rho_A)+C(\rho_A)+E(|\xi\rangle_{AC}) = P(\rho_C)+C(\rho_C)+E(|\xi\rangle_{AC}) = \alpha(d_A), \\
& P(\rho_{C'})+C(\rho_{C'})+E(|\xi\rangle_{C'B}) = P(\rho_B)+C(\rho_B)+E(|\xi\rangle_{C'B}) = \alpha(d_B).
\end{align}

As local unitary tranformation do not change entanglement, if two states are connected by such a transformation,
$$|\xi'\rangle_{AC} = U_l|\xi\rangle_{AC},\ |\eta'\rangle_{C'B} = V_l|\xi'\rangle_{C'B},$$
we have
\begin{align}
& P(\rho'_A)+C(\rho'_A)=P(\rho_A)+C(\rho_A),\ P(\rho'_C)+C(\rho'_C)=P(\rho_C)+C(\rho_C), \\
& P(\rho'_{C'})+C(\rho'_{C'})=P(\rho_{C'})+C(\rho_{C'}),\ P(\rho'_B)+C(\rho'_B)=P(\rho_B)+C(\rho_B).
\end{align}
Another way to see this is that for any subsystem, we have that for any pair of observables $X$ and $Y$ it follows that
$$P^X(\rho_s) + C^X(\rho_s) = P^Y(\rho_s) + C^Y(\rho_s) = \alpha(d_s)-E(|\Psi\rangle_{ss'}).$$

So, we can look at the predictability and coherence of local observables __sharing the eigenbasis with the local states__. Thus, in this case, the local coherence is null for all subsystems:
$$C(\xi_A)=C(\xi_C)=C(\eta_{C'})=C(\eta_B)=0.$$
But we still have
$$P(\xi_A)+E(|\xi\rangle_{AC}) = P(\xi_C)+E(|\xi\rangle_{AC})=\alpha(d_A)\ \therefore\ P(\xi_A)=P(\xi_C)=\alpha(d_A)-E(|\xi\rangle_{AC})$$
and
$$P(\eta_{C'})+E(|\eta\rangle_{C'B}) = P(\eta_B)+E(|\eta\rangle_{C'B})=\alpha(d_B)\ \therefore\ P(\eta_{C'})=P(\eta_B)=\alpha(d_B)-E(|\eta\rangle_{C'B}).$$

Remark. This result will facilitate our analysis but it also turn the results obtained in [2] obsolete.

Besides, one can see that any state with null local coherence can be transformed into the form
$$|\xi\rangle_{AC} = \sum_j c_j|jj\rangle,$$
i.e.
$$(I\otimes X(j))|\eta\rangle_{AC} = |\xi\rangle_{AC}.$$
So, without loss of generality, in what follows we shall use states of the type of $|\xi\rangle_{AC} = \sum_j c_j|jj\rangle$.
For example, for
$$|\eta\rangle_{AC} = c_0|01\rangle + c_1|10\rangle$$
we have
$$(I\otimes X)|\eta\rangle_{AC} = c_0|00\rangle + c_1|11\rangle = |\xi\rangle_{AC}.$$

## Entanglement, coherence, and predictability monotones
$l_1$-norm coherence: $$C_{l_1}(\rho_A) = \sum_{j\ne k}|\rho^A_{jk}|$$

$l_1$-norm predictability: $$P_{l_1}(\rho_A) = d-1 - \sum_{j\ne k}\sqrt{\rho^A_{jj}\rho^A_{kk}}$$

$l_1$-norm entanglement: $$E_{l_1}(|\Psi\rangle_{AB}) = \sum_{j\ne k}\big(\sqrt{\rho^A_{jj}\rho^A_{kk}} - |\rho^A_{jk}|\big)$$

$l_1$-norm complementarity: $$C_{l_1}(\rho_A) + P_{l_1}(\rho_A) + E_{l_1}(|\Psi\rangle_{AB}) = d-1.$$

## Qubits ($C_s=0$)

So, in this case, let us consider the __initial states__
$$|\xi\rangle_{AC} = \sum_{j=0}^1 c_j|jj\rangle,\ |\eta\rangle_{C'B} = \sum_{k=0}^1 d_k|kk\rangle.$$
In this case, as
$$\xi_A=\xi_C=\sum_{j=0}^1|c_j|^2|j\rangle\langle j|,\ \eta_{C'}=\eta_B=\sum_{k=0}^1|d_k|^2|k\rangle\langle k|,$$
we have that the initial predictability and entanglement are
\begin{align}
& P_{l_1}(\xi_A) = P_{l_1}(\xi_C) = 1-\sum_{j\ne k}|c_j c_k| = 1-2|c_0 c_1|,\\
& P_{l_1}(\eta_{C'}) = P_{l_1}(\eta_B) = 1-\sum_{j\ne k}|d_j d_k| = 1-2|d_0 d_1|, \\
& E_{l_1}(|\xi\rangle_{AC})= \sum_{j\ne k}|c_j c_k| = 1-P_{l_1}(\xi_C),\\
& E_{l_1}(|\eta\rangle_{C'B}) = \sum_{j\ne k}|d_j d_k| = 1 - P_{l_1}(\eta_{C'}).
\end{align}

In this case, the global state is
\begin{align}
|\Psi\rangle_{ACC'B} & = |\xi\rangle_{AC}|\eta\rangle_{C'B} \\
& = \sum_j c_j|jj\rangle_{AC}\sum_k d_k|kk\rangle_{C'B} \\
& = \sum_{j,k}c_j d_k |jk\rangle_{AB}|jk\rangle_{CC'}.
\end{align}

The Bell states in this case are given by
$$|\Phi_{pq}\rangle = \big(|p0\rangle + e^{i\pi q}|\overline{p}1\rangle\big)/\sqrt{2}.$$

The post-BBM non-normalized vectors are
\begin{align}
|\phi_{pq}\rangle_{ABCC'} & = \big(I_{AB}\otimes|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'}\big)|\Psi\rangle_{ACC'B} \\
& = \sum_{j,k}c_j d_k |jk\rangle_{AB}\otimes|\Phi_{pq}\rangle_{CC'}\langle\Phi_{pq}|jk\rangle_{CC'} \\
& = \sum_{j,k}c_j d_k \frac{1}{\sqrt{2}}\big(\delta_{pj}\delta_{0k} + e^{-i\pi q}\delta_{\overline{p}j}\delta_{1k}\big) |jk\rangle_{AB}\otimes|\Phi_{pq}\rangle_{CC'} \\
& = \frac{1}{\sqrt{2}}\big(c_p d_0|p0\rangle + e^{-i\pi q}c_\overline{p} d_1|\overline{p}1\rangle\big)\otimes|\Phi_{pq}\rangle_{CC'}.
\end{align}

Norm 
$$||(\phi_{pq})_{AB}||^2 = \frac{1}{2}\big(|c_p d_0|^2 + |c_{\overline{p}}d_1|^2\big) = Pr(\Phi_{pq}^{CC'}),$$
as will be shown below.

Post-BBM state
\begin{align}
|\hat{\phi}_{pq}\rangle_{AB} & = \frac{|\phi_{pq}\rangle_{AB}}{||(\phi_{pq})_{AB}||} = \frac{1}{\sqrt{2Pr(\Phi_{pq}^{AB})}}\big(c_p d_0|p0\rangle + e^{-i\pi q}c_\overline{p} d_1|\overline{p}1\rangle\big)
\end{align}

Post-BBM reduced density operator
\begin{align}
\big(\hat{\phi}_{pq}\big)_{A} & = \frac{1}{2Pr(\Phi_{pq}^{AB})}\big(|c_p d_0|^2|p\rangle\langle p| + |c_\overline{p} d_1|^2|\overline{p}\rangle\langle \overline{p}|\big)
\end{align}

Using  the $l_1$-norm entanglement, $E_{l_1}(|\psi\rangle_{AB}) = \sum_{j\ne k}\big(\sqrt{\psi^A_{jj}\psi^A_{kk}} - |\psi^A_{jk}|\big)$, we get
\begin{align}
E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) & = 2\sqrt{\big(\hat{\phi}_{pq}\big)^{A}_{pp}\big(\hat{\phi}_{pq}\big)^{A}_{\overline{p},\overline{p}}} = 2\frac{\sqrt{|c_p d_0|^2|c_\overline{p} d_1|^2}}{2Pr(\Phi_{pq}^{AB})} \\
& = \frac{2|c_p c_\overline{p}|\times 2|d_0 d_1|}{4Pr(\Phi_{pq}^{AB})} \\
& = \frac{E_{l_1}(|\xi\rangle_{AC})E_{l_1}(|\eta\rangle_{C'B})}{4Pr(\Phi_{pq}^{AB})} \\
& = \frac{\big(1-P_{l_1}(\xi_C)\big)\big(1-P_{l_1}(\eta_{C'})\big)}{4Pr(\Phi_{pq}^{AB})} \\
\end{align}

## Probabilities

For the preparation 
$$|\Psi\rangle_{ABCC'} = \sum_{j,k=0}^{d-1}c_j d_k|jk\rangle_{AB}|jk\rangle_{CC'},$$
the probability of obtaining a given result in the BBM is
\begin{align}
Pr(\Phi_{pq}^{CC'}) & = Pr(\Phi_{pq}^{CC'}|\Psi_{ABCC'}) \\
& = Tr_{CC'}(\rho_{CC'}|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'}) \\
& = Tr_{CC'}(Tr_{AB}(|\Psi\rangle\langle\Psi|_{ABCC'})|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'}) \\
& = Tr_{ABCC'}(|\Psi\rangle\langle\Psi|_{ABCC'}(I_{AB}\otimes|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'})) \\
& = \langle\Psi|(I_{AB}\otimes|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'})|\Psi\rangle_{ABCC'}
\end{align}

### qubits ($C_s=0$)
\begin{align}
Pr(\Phi_{pq}^{CC'}) & = \langle\Psi|(I_{AB}\otimes|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'})|\Psi\rangle_{ABCC'} \\
& = (\sum_{j,k}c_j d_k\langle jk|_{AB}\langle jk|_{CC'})(I_{AB}\otimes\frac{1}{\sqrt{2}}(|p0\rangle+e^{i\pi q}|\overline{p}1\rangle)\frac{1}{\sqrt{2}}(\langle p0|+e^{-i\pi q}\langle\overline{p}1|)_{CC'})\sum_{l,m}c_l d_m|lm\rangle_{AB}|lm\rangle_{CC'} \\
& = \frac{1}{2}\sum_{jklm}c_j d_k c_l d_m\langle jk|lm\rangle_{CC'}\langle jk|(|p0\rangle+e^{i\pi q}|\overline{p}1\rangle)_{AB}(\langle p0|+e^{-i\pi q}\langle\overline{p}1|)|lm\rangle_{CC'} \\
& = \frac{1}{2}\sum_{jklm}c_j d_k c_l d_m \delta_{jl}\delta_{km}\langle jk|(|p0\rangle+e^{i\pi q}|\overline{p}1\rangle)_{AB}(\langle p0|+e^{-i\pi q}\langle\overline{p}1|)|lm\rangle_{CC'} \\
& = \frac{1}{2}\sum_{jk}c_j^2 d_k^2(\delta_{jp}\delta_{k0}+e^{i\pi q}\delta_{j,\overline{p}}\delta_{k1})(\delta_{jp}\delta_{k0}+e^{-i\pi q}\delta_{j,\overline{p}}\delta_{k,1}) \\
& = \frac{1}{2}\sum_{jk}c_j^2 d_k^2(\delta_{jp}\delta_{k0}+\delta_{j,\overline{p}}\delta_{k1}) \\
& = \frac{1}{2}(c_p^2 d_0^2 + c_\overline{p}^2 d_1^2).
\end{align}

So
\begin{align}
& Pr(\Phi_{00}^{CC'}) = Pr(\Phi_{01}^{CC'}) = \frac{1}{2}(c_0^2 d_0^2 + c_1^2 d_1^2), \\
& Pr(\Phi_{10}^{CC'}) = Pr(\Phi_{11}^{CC'}) = \frac{1}{2}(c_1^2 d_0^2 + c_0^2 d_1^2),
\end{align}
recalling that 
$$\Phi_{00} = \Phi_+,\ \Phi_{01} = \Phi_-,\ \Phi_{10} = \Psi_+,\ \Phi_{11}=-\Psi_-.$$

### qudits  ($C_s=0$)
\begin{align}
Pr(|\Phi_{pq}\rangle_{CC'}) & = \langle\Psi|(I_{AB}\otimes|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'})|\Psi\rangle_{ABCC'} \\
& = (\sum_{j,k=0}^{d-1}c_j d_k\langle j,k|_{AB}\langle j,k|_{CC'})(I_{AB}\otimes\frac{1}{\sqrt{d}}\sum_{n=0}^{d-1}e^{2\pi iqn/d}|p\oplus n,n\rangle_{CC'}\frac{1}{\sqrt{d}}\sum_{o=0}^{d-1}e^{-2\pi iqo/d}\langle p\oplus o,o|_{CC'})\sum_{l,m=0}^{d-1}c_l d_m|l,m\rangle_{AB}|l,m\rangle_{CC'} \\
& = \frac{1}{d}\sum_{j,k,l,m,n,o=0}^{d-1}c_j d_k c_l d_m e^{2\pi iq(n-o)/d}\langle j,k|l,m\rangle_{AB}\langle j,k|p\oplus n,n\rangle_{CC'}\langle p\oplus o,o|l,m\rangle_{CC'} \\
& = \frac{1}{d}\sum_{j,k,l,m,n,o=0}^{d-1}c_j d_k c_l d_m e^{2\pi iq(n-o)/d}\delta_{j,l}\delta_{k,m}\langle j,k|p\oplus n,n\rangle_{CC'}\langle p\oplus o,o|l,m\rangle_{CC'} \\
& = \frac{1}{d}\sum_{j,k,n,o=0}^{d-1}c_j^2 d_k^2 e^{2\pi iq(n-o)/d}\langle j,k|p\oplus n,n\rangle_{CC'}\langle p\oplus o,o|j,k\rangle_{CC'} \\
& = \frac{1}{d}\sum_{j,k,n,o=0}^{d-1}c_j^2 d_k^2 e^{2\pi iq(n-o)/d}\delta_{j,p\oplus n}\delta_{k,n}\delta_{p\oplus o,j}\delta_{o,k} \\
& = \frac{1}{d}\sum_{j,k=0}^{d-1}c_j^2 d_k^2 e^{2\pi iq(k-k)/d}\delta_{j,p\oplus k}\delta_{p\oplus k,j} \\
& = \frac{1}{d}\sum_{j,k=0}^{d-1}c_j^2 d_k^2\delta_{j,p\oplus k} \\
& = \frac{1}{d}\sum_{k=0}^{d-1}c_{p\oplus k}^2 d_k^2
\end{align}

## Qudits ($C_s =0$)

Let us consider the __initial states__
$$|\xi\rangle_{AC} = \sum_{j=0}^{d-1} c_j|jj\rangle;\ |\eta\rangle_{C'B} = \sum_{k=0}^{d-1} d_k|kk\rangle,$$
with $c_j\in\mathbb{R}.$ In this case, as
$$\xi_A=\xi_C=\sum_{j=0}^{d-1}|c_j|^2|j\rangle\langle j|;\  \eta_{C'}=\eta_B=\sum_{k=0}^{d-1}|d_k|^2|k\rangle\langle k|,$$
we have that the initial predictability and entanglement are
\begin{align}
& P_{l_1}(\xi_A) = P_{l_1}(\xi_C) = d-1-\sum_{j\ne k}|c_j c_k|;\ P_{l_1}(\eta_{C'}) = P_{l_1}(\eta_B) = d-1-\sum_{j\ne k}|d_j d_k|,\\
& E_{l_1}(|\xi\rangle_{AC}) = \sum_{j\ne k}|c_j c_k|;\ E_{l_1}(|\eta\rangle_{C'B}) = \sum_{j\ne k}|d_j d_k|.
\end{align}
Again 
$$C_{l_1}(\rho_s)=0$$
for $s=A,B,C,C'.$

The global state is
\begin{align}
|\Psi\rangle_{ACC'B} & = |\xi\rangle_{AC}|\eta\rangle_{C'B} = \sum_{j=0}^{d-1} c_j|jj\rangle_{AC}\sum_{k=0}^{d-1} d_k|kk\rangle_{C'B} = \sum_{j,k=0}^{d-1}c_j d_k |jk\rangle_{AB}|jk\rangle_{CC'}.
\end{align}

The Bell states are
\begin{align}
|\Phi_{pq}\rangle & = (W_{pq}\otimes I)|\Phi\rangle  \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{2\pi iqj/d}|(p+j)\bmod d\rangle\otimes|j\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{j=0}^{d-1}e^{2\pi iqj/d}|p\oplus j\rangle\otimes|j\rangle.
\end{align}

The post-BBM non-normalized vectors are
\begin{align}
|\phi_{pq}\rangle_{ABCC'} & = \big(I_{AB}\otimes|\Phi_{pq}\rangle\langle\Phi_{pq}|_{CC'}\big)|\Psi\rangle_{ACC'B} \\
& = \sum_{j,k=0}^{d-1}c_j d_k |jk\rangle_{AB}\otimes|\Phi_{pq}\rangle_{CC'}\langle\Phi_{pq}|jk\rangle_{CC'} \\
& = \sum_{j,k=0}^{d-1}c_j d_k \frac{1}{\sqrt{d}}e^{-2\pi iqk/d}\delta_{j,p\oplus k} |jk\rangle_{AB}\otimes|\Phi_{pq}\rangle_{CC'} \\
& = \Big(\frac{1}{\sqrt{d}}\sum_{k=0}^{d-1}c_{p\oplus k} d_k e^{-2\pi iqk/d}|p\oplus k,k\rangle_{AB}\Big)\otimes|\Phi_{pq}\rangle_{CC'},
\end{align}
where we used
\begin{align}
\langle\Phi_{pq}|jk\rangle & = \Big(\frac{1}{\sqrt{d}}\sum_{l=0}^{d-1}e^{-2\pi iql/d}\langle p\oplus l|\otimes\langle l|\Big)|jk\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{l=0}^{d-1}e^{-2\pi iql/d}\langle p\oplus l|j\rangle\otimes\langle l|k\rangle \\
& = \frac{1}{\sqrt{d}}\sum_{l=0}^{d-1}e^{-2\pi iql/d}\langle p\oplus l|j\rangle\otimes \delta_{l,k} \\
& = \frac{1}{\sqrt{d}}e^{-2\pi iqk/d}\langle p\oplus k|j\rangle \\
& = \frac{1}{\sqrt{d}}e^{-2\pi iqk/d}\delta_{j,p\oplus k}.
\end{align}

For Alice and Bob, the post-BBM non-normalized vector is
$$|\phi_{pq}\rangle_{AB} = \frac{1}{\sqrt{d}}\sum_{k=0}^{d-1}c_{p\oplus k} d_k e^{-2\pi iqk/d}|p\oplus k,k\rangle_{AB}.$$

Norm
\begin{align}
|||\phi_{pq}\rangle_{AB}||^2 & = \langle\phi_{pq}|\phi_{pq}\rangle_{AB} \\
& = \frac{1}{\sqrt{d}}\sum_{k=0}^{d-1}c_{p\oplus k} d_k e^{2\pi iqk/d}\langle p\oplus k,k|_{AB}\frac{1}{\sqrt{d}}\sum_{l=0}^{d-1}c_{p\oplus l} d_l e^{-2\pi iql/d}|p\oplus l,l\rangle_{AB} \\
& = \frac{1}{d}\sum_{k,l=0}^{d-1}c_{p\oplus k} d_k c_{p\oplus l} d_l e^{2\pi iq(k-l)/d}\langle p\oplus k,k|p\oplus l,l\rangle_{AB} \\
& = \frac{1}{d}\sum_{k,l=0}^{d-1}c_{p\oplus k} d_k c_{p\oplus l} d_l e^{2\pi iq(k-l)/d}\delta_{k,l} \\
& = \frac{1}{d}\sum_{k=0}^{d-1}c_{p\oplus k} d_k c_{p\oplus k} d_k e^{2\pi iq(k-k)/d}\delta_{k,l} \\
& = \frac{1}{d}\sum_{k=0}^{d-1}c_{p\oplus k}^2 d_k^2 \\
& = Pr(|\Phi_{pq}\rangle_{CC'}).
\end{align}

This is expected because one can always write
\begin{align}
|\Psi\rangle_{ABCC'} & = \sum_{p,q} |||\phi_{pq}\rangle_{AB}||\frac{|\phi_{pq}\rangle_{AB}}{|||\phi_{pq}\rangle_{AB}||}|\Phi_{pq}\rangle_{CC'} \\
& = \sum_{p,q} \sqrt{Pr(\Phi_{pq}^{CC'})}|\hat{\phi}_{pq}\rangle_{AB}|\Phi_{pq}\rangle_{CC'} \\
\end{align}

Normalizing we get the post-BBM pure state
\begin{align}
|\hat{\phi}_{pq}\rangle_{AB} & = \frac{|\phi_{pq}\rangle_{AB}}{|||\phi_{pq}\rangle_{AB}||} \\
& = \frac{1}{\sqrt{ Pr(|\Phi_{pq}\rangle_{CC'}) d}}\sum_{k=0}^{d-1}c_{p\oplus k} d_k e^{2\pi iqk/d}|p\oplus k\rangle_A\otimes|k\rangle_B.
\end{align}

Related density operator
\begin{align}
\hat{\phi}_{pq}^{AB} & = |\hat{\phi}_{pq}\rangle\langle\hat{\phi}_{pq}|_{AB} \\
& = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{k=0}^{d-1}c_{p\oplus k} d_k e^{2\pi iqk/d}|p\oplus k\rangle_A\otimes|k\rangle_B \sum_{l=0}^{d-1}c_{p\oplus l} d_l e^{-2\pi iql/d}\langle p\oplus l|_A\otimes\langle l|_B \\
& = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{k,l=0}^{d-1}c_{p\oplus k}c_{p\oplus l} d_k d_l e^{2\pi iq(k-l)/d}|p\oplus k\rangle\langle p\oplus l|_A\otimes|k\rangle\langle l|_B \\
\end{align}

Reduced density operator
\begin{align}
\hat{\phi}_{pq}^{A} & = Tr_B(\hat{\phi}_{pq}^{AB}) \\
& = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{k,l=0}^{d-1}c_{p\oplus k}c_{p\oplus l} d_k d_l e^{2\pi iq(k-l)/d}|p\oplus k\rangle\langle p\oplus l|_A\otimes Tr_B(|k\rangle\langle l|_B) \\
& = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{k,l=0}^{d-1}c_{p\oplus k}c_{p\oplus l} d_k d_l e^{2\pi iq(k-l)/d}|p\oplus k\rangle\langle p\oplus l|_A \delta_{l,k} \\
& = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{k=0}^{d-1}c_{p\oplus k}^2 d_k^2|p\oplus k\rangle\langle p\oplus k|_A.
\end{align}

$l_1$-norm entanglement
\begin{align}
E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) & = \sum_{j\ne k}\Big(\sqrt{(\hat{\phi}_{pq}^{A})_{jj}(\hat{\phi}_{pq}^{A})_{kk}} - |(\hat{\phi}_{pq}^{A})_{jk}|\Big) \\
& = \sum_{j\ne k}\Big(\sqrt{\frac{c_{p\oplus j}^2 d_j^2 }{Pr(|\Phi_{pq}\rangle_{CC'}) d}\frac{c_{p\oplus k}^2 d_k^2 }{Pr(|\Phi_{pq}\rangle_{CC'}) d}} - 0\Big) \\
& = \sum_{j\ne k}\frac{|c_{p\oplus j} c_{p\oplus k} d_j d_k|}{Pr(|\Phi_{pq}\rangle_{CC'}) d} \\
& = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k|.
\end{align}

### Special cases

* If the initial local __predictability is maximum__ (minimum initial entanglement), i.e. $c_j=1,\ d_k=1$ for some $j,k$, then all other coefficients are null and therefore the final entanglement is null:
$$P_{l_1}(\xi_A)=1\ \text{or}\ P_{l_1}(\eta_B)=1\ \therefore\ E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB})=0.$$

* If the initial __predictability is null__ (maximum initial entanglement), i.e. $c_j=\frac{1}{d}\ \forall j\ \text{and}\ d_k=\frac{1}{d}\ \forall k$, then we have
\begin{align}
Pr(|\Phi_{pq}\rangle_{CC'}) & = \frac{1}{d}\sum_{j=0}^{d-1}c_{p\oplus j}^2d_j^2 = \frac{1}{d}\sum_{j=0}^{d-1}\frac{1}{d^2}\frac{1}{d^2} = \frac{1}{d}d\frac{1}{d^2}\frac{1}{d^2} = \frac{1}{d^4},
\end{align}
and the final entanglement is maximum:
\begin{align}
E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) & = \sum_{j\ne k}\frac{|c_{p\oplus j} c_{p\oplus k} d_j d_k|}{Pr(|\Phi_{pq}\rangle_{CC'}) d} = \sum_{j\ne k}\frac{|(1/d)(1/d)(1/d)(1/d)|}{\frac{1}{d^4} d} = \sum_{j\ne k}\frac{1}{d} = \frac{d(d-1)}{d} \\
& = d-1.
\end{align}

These results are as expected.

* Another indication that the result is correct is the special case of __qubits__. For any 
$$(\forall q,\ p=0)$$
we shall have
\begin{align}
E_{l_1}(|\hat{\phi}_{0q}\rangle_{AB}) & = \sum_{j\ne k}\frac{|c_{0\oplus j} c_{0\oplus k} d_j d_k|}{Pr(|\Phi_{pq}\rangle_{CC'}) d} = \frac{|c_0 c_1 d_0 d_1|+|c_1 c_0 d_1 d_0|}{2Pr(|\Phi_{pq}\rangle_{CC'})} = \frac{2|c_0 c_1 d_0 d_1|}{2Pr(|\Phi_{pq}\rangle_{CC'})}
\end{align}
and
\begin{align}
E(|\xi\rangle_{AC})E(|\eta\rangle_{C'B}) & = \sum_{j\ne k}|c_j c_k|\sum_{l\ne m}|d_l d_m| = (|c_0 c_1|+|c_1 c_0|)(|d_0 d_1|+|d_1 d_0|) = 4|c_0 c_1 d_0 d_1|.
\end{align}
Therefore
$$E_{l_1}(|\hat{\phi}_{0q}\rangle_{AB}) = \frac{E(|\xi\rangle_{AC})E(|\eta\rangle_{C'B})}{4Pr(|\Phi_{1q}\rangle_{CC'})},$$
as we have obtained before.

* Besides, for any 
$$(\forall q,\ p=1)$$
we shall have
\begin{align}
E_{l_1}(|\hat{\phi}_{1q}\rangle_{AB}) & = \sum_{j\ne k}\frac{|c_{1\oplus j} c_{1\oplus k} d_j d_k|}{Pr(|\Phi_{pq}\rangle_{CC'}) d} = \frac{|c_1 c_0 d_0 d_1|+|c_0 c_1 d_1 d_0|}{2Pr(|\Phi_{pq}\rangle_{CC'})} = \frac{2|c_0 c_1 d_0 d_1|}{2Pr(|\Phi_{pq}\rangle_{CC'})}.
\end{align}
Therefore
$$E_{l_1}(|\hat{\phi}_{1q}\rangle_{AB}) = \frac{E(|\xi\rangle_{AC})E(|\eta\rangle_{C'B})}{4Pr(|\Phi_{1q}\rangle_{CC'})},$$
as we have obtained before.

### Summary of equations

$$|\xi\rangle_{AC} = \sum_{j=0}^{d-1}c_j|jj\rangle,\ |\eta\rangle_{C'B} = \sum_{k=0}^{d-1}d_k|kk\rangle$$

$$P(\xi_A) = d-1-\sum_{j\ne k}|c_j c_k|,\ P(\eta_B) = d-1-\sum_{j\ne k}|d_j d_k|$$

$$E_{l_1}(|\xi\rangle_{AC}) = \sum_{j\ne k}|c_j c_k|,\ E_{l_1}(|\eta\rangle_{C'B}) = \sum_{j\ne k}|d_j d_k|$$ 

$$|\hat{\phi}_{pq}\rangle_{AB} = \frac{1}{\sqrt{ Pr(|\Phi_{pq}\rangle_{CC'}) d}}\sum_{k=0}^{d-1}c_{p\oplus k} d_k e^{2\pi iqk/d}|p\oplus k\rangle_A\otimes|k\rangle_B$$

$$ Pr(\Phi_{pq}^{CC'}) = \frac{1}{d}\sum_{j=0}^{d-1}c_{p\oplus j}^2d_j^2$$

$$E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) = \frac{1}{Pr(|\Phi_{pq}\rangle_{CC'}) d}\sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| $$

### Two-qutrit case

In this case $p,q=0,1,2$ and
\begin{align}
Pr(\Phi_{pq}^{CC'}) & = \frac{1}{3}\sum_{j=0}^{3-1}c_{p\oplus j}^2d_j^2 = \frac{1}{3}(c_{p}^2 d_0^2+c_{p\oplus 1}^2 d_1^2+c_{p\oplus 2}^2 d_2^2)
\end{align}

\begin{align}
\sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| & = 2\sum_{j< k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| \\
& = 2(|c_{p\oplus 0} c_{p\oplus 1} d_0 d_1| + |c_{p\oplus 0} c_{p\oplus 2} d_0 d_2| + |c_{p\oplus 1} c_{p\oplus 2} d_1 d_2|)
\end{align}.

Let us notice that

\begin{align}
\sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| &  = 2 \sum_p(|c_{p\oplus 0} c_{p\oplus 1} d_0 d_1| + |c_{p\oplus 0} c_{p\oplus 2} d_0 d_2| + |c_{p\oplus 1} c_{p\oplus 2} d_1 d_2|) \\ & = 2\left(|d_0 d_1|  \sum_p |c_{p\oplus 0} c_{p\oplus 1}| + |d_0 d_2| \sum_p |c_{p\oplus 0} c_{p\oplus 2}| + |d_1 d_2| \sum_p |c_{p\oplus 1} c_{p\oplus 2}|\right) \\& = 2 \left(|d_0 d_1| + |d_1 d_2| + |d_0 d_2|\right) \frac{1}{2}E_{l_1}(|\xi\rangle_{AC}) \\
& = \frac{1}{2}E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B}),
\end{align}
which implies that
\begin{align}
\sum_p Pr(\Phi_{pq}^{CC'}) E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) = \frac{1}{d} \sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| = \frac{1}{2d} E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B}),
\end{align}
and the average over all post-measurement entangled states
\begin{align}
\langle E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) \rangle = \sum_q \sum_p Pr(\Phi_{pq}^{CC'}) E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) = \sum_ q \frac{1}{d} \sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| = \frac{1}{2} E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B})
\end{align}
Moreover, let us notice that
\begin{align}
\sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| \le \sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| = \frac{1}{2}E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B}),
\end{align}
and, therefore,
$$E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) \le \frac{1}{2 Pr(|\Phi_{pq}\rangle_{CC'}) d}E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B}) $$




### General case

The general case is the same as the qutrit case. To see this, let us notice that

$$ \sum_p Pr(\Phi_{pq}^{CC'}) = \frac{1}{d}\sum_{j=0}^{d-1}d_j^2 (\sum_p c_{p\oplus j}^2) = \frac{1}{d}\sum_{j=0}^{d-1}d_j^2 = \frac{1}{d}$$ 
and, therefore,
$$ \sum_q \sum_p Pr(\Phi_{pq}^{CC'}) = 1.$$
Moreover, let us notice that
\begin{align}
\sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k|= \sum_{j\ne k}|d_j d_k| (\sum_p |c_{p\oplus j} c_{p\oplus k}|), 
\end{align}
however, in this case, the sum over does not imply that $\frac{1}{2}E_{l_1}(|\xi\rangle_{AC})$, once the the sum over $p$ does not account for all the terms contained in $E_{l_1}(|\xi\rangle_{AC})$.

However, the following inequality it is true
\begin{align}
\sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| &= \sum_{j\ne k}|d_j d_k| (\sum_p |c_{p\oplus j} c_{p\oplus k}|)\\
& \le  \sum_{j\ne k}|d_j d_k| \sum_{l\neq m} |c_l c_m|,\\
& = E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B}) 
\end{align}
once $j \neq k$, then $p\oplus j \neq p \oplus k$. By defining $l = p\oplus j$ and $m = p \oplus k $, we have $l \neq m$. However, the sum over $p$ contains $d-1$ terms while the sum over $l \neq m$ contains $d(d-1)$ terms.

Hence,
\begin{align}
\langle E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) \rangle = \sum_q \sum_p Pr(\Phi_{pq}^{CC'}) E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) = \sum_ q \frac{1}{d} \sum_p \sum_{j\ne k}|c_{p\oplus j} c_{p\oplus k} d_j d_k| \le E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B})
\end{align}
and
$$E_{l_1}(|\hat{\phi}_{pq}\rangle_{AB}) \le \frac{1}{ Pr(|\Phi_{pq}\rangle_{CC'}) d}E_{l_1}(|\xi\rangle_{AC}) E_{l_1}(|\eta\rangle_{C'B}) $$





```python
import numpy as np

# Define d = 4
d = 4

# Example vectors with real, positive entries
c = np.array([1.0, 2.0, 3.0, 4.0])
d_vec = np.array([1.5, 2.5, 0.5, 1.0])

# E_l1 entanglements
def E_l1(x):
    return sum([abs(x[j]*x[k]) for j in range(d) for k in range(d) if j != k])

E_c = E_l1(c)
E_d = E_l1(d_vec)

# Full double sum with shifts
S = 0
for p in range(d):
    for j in range(d):
        for k in range(d):
            if j != k:
                idx_j = (p + j) % d
                idx_k = (p + k) % d
                S += abs(c[idx_j] * c[idx_k] * d_vec[j] * d_vec[k])

# Compute ratio
ratio = S / (E_c * E_d)

print(f"E_l1(c) = {E_c:.4f}")
print(f"E_l1(d) = {E_d:.4f}")
print(f"S = {S:.4f}")
print(f"Ratio = S / (E_c * E_d) = {ratio:.4f}")
```
